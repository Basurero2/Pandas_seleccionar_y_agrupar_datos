## Analisis de las emisiones

Vamos a realizar un analisis, pero esta vez, de la cantidad de la emision total por tipo de gas. Para esto, vamos a realizar unb agrupamiento de los datos con la función groupby()

```python
emisiones_por_año.groupby("Gás").groups
#muestra un diccionario de valores de la Gás
#donde el valor de cada uno es una lista donde los elementos son los indices donde aparece
#este valor de Gás en el DataFrame
```

```python
emisiones_por_año.groupby("Gás").get_group("CO2 (t)").head()
#mostrar el DataFrame con todos los registros donde este el valor "CO2 (t)"
#en la columna "Gás"
#agrupo los valores de "Gas" y luego accedio a uno "CO2 (t)"
```

Metodos posibles para groupby()
- count() : cuantos elementos hubo para agrupacion de cada elemento (repeticiones)
. sum() : suma de valores
- mean() : promedio
- median() : medianba
- min() : valor minimo
- max() : valor maximo
- std() : desviacion estandar (raiz cuadrada de varianza)
- var() : varianza



Vamos a sumar todas las emisiones de los años 1970 hasta 2021 (ya fue transpuesto esas columnas como registros para asosciarlo a cada valor con tipo de gas mejor) de cada tipo de gas

```python
emisiones_por_año.groupby("Gás").sum(numeric_only=True)
#no especificamos con que columna se va asociar esos valores a sumar
#por eso se pone numeric_only True, para indicar que solo numeros se van a operar
#la unica columna en nuestro caso es Emision con esos datos
```

```python
emisiones_por_año.groupby("Gás")["Emision"].sum()
#este es otra forma, especificando la columna para que opere con sus valores
```

Vamos a ordenar los datos de mayor a menor

```python
emisiones_por_año.groupby("Gás").sum(numeric_only=True).sort_values("Emision", ascending=False)
#aqui se especifica la columna a ordenar
#yaque al no especifica con que columna se va a operar el sum()
#pues toma la columna donde tenga valores numericos
#por lo tanto en sort_values se necesita especificar
```

```python
emisiones_por_año.groupby("Gás")["Emision"].sum().sort_values(ascending=False)
#esto devuleve una serie, donde no se especifica que columna se va a ordena
#yaque solo existe 1 en una serie, y ya especificamos con que columna trabajar en groupby
```

```python
emisiones_por_gas = emisiones_por_año.groupby("Gás")["Emision"].sum().sort_values(ascending=False)
```

Creamos una grafica para analizar mejor esos valores en notacion cientificay ver comparacion de cantidad de emisiones por gas del año 1970 hasta 2021

```python
emisiones_por_gas.plot(kind="barh", figsize=(10,6))
```

Viendo la grafica vemos que el tipo de gas CO2 es el que mas abunda, por lo tanto vamos a analizar el pórcentaje de las emisiones de todos los tipos de C02 con todas las emisiones de todos los gases del año 1970 hasta 2021

```python
float((emisiones_por_gas.iloc[0:9].sum() / emisiones_por_gas.sum()) * 100)
#iloc : muestra una serie o DataFrame (2 args), donde mostrara filas o columnas segun su indice (arg)
```

```python
print(f"Las emisiones de CO2 corresponden al {float((emisiones_por_gas.iloc[0:9].sum() / emisiones_por_gas.sum()) * 100):.2f}% de las emisiones totales de gases de efecto invernadero en Brasil de 1970 a 2021")
```

### 1. Descubrir cuales son los sectores mas contaminantes por cada pais

```python
emisiones_por_año.head()
```

Vamos a mostrar una serie multiindex con agrupacion con groupby()
Harmeos 2 versiones para ver el orden si importa

```python
gas_por_sector = emisiones_por_año.groupby(["Gás", "Nível 1 - Setor"])["Emision"].sum()
gas_por_sector
```

```python
sector_por_gas = emisiones_por_año.groupby(["Nível 1 - Setor", "Gás"])["Emision"].sum()
sector_por_gas
```

Ejemplo : vamos a ver que sector emite mayor cantidad de CO2 (t)
- usaremos el metodo cross section : xs()

```python
gas_por_sector.xs("CO2 (t)", level=0)
#de la serie se va a seleccionar del primer indice (level=0) de los indices
#el valor agrupado de "CO2 (t)"
```

```python
gas_por_sector.xs(("CO2 (t)", "Mudança de Uso da Terra e Floresta"), level=[0,1])
#seleccionar de indice 0 el valor agrupado de C02 (t)
#seleccionar tambien indice 1 valor agrupado de Mudança de Uso da Terra e Floresta
#claramente ese valor de indice 1 esta dentro de "CO2 (t)" (que esta en indice 0)
#resultado : una serie de un registro
```

```python
gas_por_sector.xs("CO2 (t)", level=0).max()
#muestra el valor maximo (columna) de los registros que estan dentro del valor agrupado "C02 (t)"
#la cual esta en el indice 0
```

```python
gas_por_sector.xs("CO2 (t)", level=0).idxmax()
#muestra el valor asociado (indice) al valor maximo (columna) de los registros dentro
#del valor agrupado "C02 (t)" la cual esta en el indice 0
```

Para evitar un metodo manual para cada tipo de gas como el caso anterior que usamos solo para el tipo "CO2 (t)", usaremos una forma mas general usando **groupby()**, **max()** y **idxmax()**

```python
gas_por_sector.groupby(level=0).max()
#nos dara el valor de emison mas alta (dado por un sector) por cada tipo de gas
#pero no tenemos el nombre del sector asociado a ese numero
#explicacion :
#   al no poner un valor del level = 0, nos muestra todos los valores
#  max() : busca el valor maximo (columna)
#  valores de columna Emision seran esos valores maximos
```

```python
gas_por_sector.groupby(level=0).idxmax()
#aqui obtenemos el nombre del sector que emite mas cantidad para cada tipo de gas
#explicacion :
#   al no poner un valor del level = 0, nos muestra todos los valores
#  idxmax() : busca el valor maximo (columna) pero nos muestra el valor de indice (level 1) asociado
#  valores de columna Emision seran esos valores del indice level 1 (ordern : level 0 - 1 - 2 - etc)
```

Como vemos necesitados de ambas series los datos de Emisiones en 2 columnas  para ello crearemos un DataFrame añadiendo 2 columnas con esos datos con el metodo **insert()**
- usaremos como base la serie usada con **idxmax()** y se le añadira los datos de la columna Emision de la serie usada con **max()**

```python
valores_max = gas_por_sector.groupby(level=0).max().values
valores_max
```

```python
tabla_totalizada = gas_por_sector.groupby(level=0).idxmax().to_frame()
tabla_totalizada.insert(1, "Cantidad de emision", valores_max)
tabla_totalizada
#para usar insert necesita ser un DataFrame
#por eso a la serie se puso .to_frame() para convertir a DataFrame
```

```python
tabla_totalizadas = pd.DataFrame({
    "Sector Mayor Emisión": gas_por_sector.groupby(level=0).idxmax(),
    "Cantidad de emision": gas_por_sector.groupby(level=0).max()
})
#otra forma de convertir a DataFrame, coloca las 2 series en orden, se conserva todo
```

####  resultado : Descubrir cuales son los sectores mas contaminantes por cada pais

```python
tabla_totalizadas
```

### 2. Descubrir cuales son los gases mas contaminantes por sector

Anteriormente usamos otro orden para la serie multindex en la variable **sector_por_gas**, pero usaremos un metodo que hara lo mismo : **swaplevel()**

```python
sector_por_gas
```

Usaremos el swaplevel() en variable sector_por_gas para cambiar de posiciones los indices

```python
gas_por_sector.swaplevel(0, 1)
#el indice 0 se va a cambiar con el indice 1 de posicion
```

#### resultado : Descubrir cuales son los gases mas contaminantes por sector

Ahora que lo tenemos agrupados correctamente usaremos el **idxmax()** para busque los valores maximos pero muestre los valores asociados al indice 1
- aqui no se añadio una columna de los valores numericos yaque solo nos pedian el nombre del gas que emitia mas por cada sector, pero se le puede añadir usanda la misma logica que el primer pedido

```python
gas_por_sector.swaplevel(0, 1).groupby(level=0).idxmax()
```

Ahora usaremos **pivot_table()**, para crear una tabla dinamica, osea vamos a transofrmar una tabla formato Long a Wide, usaremos tres parametros
